```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
import cdflib
from scipy.fft import fft
from matplotlib.gridspec import GridSpec
from datetime import datetime
```

```python
def movingaverage(interval, window_size):
    window= np.ones(int(window_size))/float(window_size)
    return np.convolve(interval, window, 'same')

def extract_date(start_date):
    return start_date.strftime('%Y%m%d')
```

```python
v = 389 * 1000 # m/s
n_p = 3.71 * 10**6 # m-3
mp = 1.6726219 * 10**-27 # kilograms
mu_0 = 1.25663706 * 10**-6 # m kg s-2 A-2
```

```python
df_dates = pd.read_csv(r"C:\Users\urvin\Downloads\fast-wind-catalog.csv")
df_dates['start_date'] = pd.to_datetime(df_dates['start_date'], format = '%d/%m/%Y')
df_dates['end_date'] = pd.to_datetime(df_dates['end_date'], format = '%d/%m/%Y') 
```

```python
my_folder2 = r"C:\DATA\SOLO\Data\psd\slow"
```

```python
save_path = r"C:\DATA\SOLO\plots\psd\slow"
```

```python
for index, row in df_dates.iterrows():
    start_date = row['start_date']
    date_folder_name = f'SOLO{extract_date(start_date)}'
        
    # Check if the folder exists
    full_folder_path = os.path.join(my_folder2, date_folder_name)
    if os.path.exists(full_folder_path):
        # Get a list of CDF files in the folder
        file_pattern = os.path.join(full_folder_path, '*.cdf')
        cdf_files = [file for file in os.listdir(full_folder_path) if file.endswith('.cdf')]
        data_frames = []
        for file_name in cdf_files:
            full_file_name = os.path.join(full_folder_path, file_name)
            print(f"Now reading {full_file_name}")

            with cdflib.CDF(full_file_name) as cdf:
                epoch_cdf = cdf['Epoch'][:]
                epoch_datetime = pd.to_datetime(cdflib.cdfepoch.to_datetime(epoch_cdf))
                data = {
                    'Epoch': epoch_datetime,
                    'Bx': cdf['B_RTN'][:, 0],
                    'By': cdf['B_RTN'][:, 1],
                    'Bz': cdf['B_RTN'][:, 2]
                }
                dat = pd.DataFrame(data)
                data_frames.append(dat)
        
        df_mag = pd.concat(data_frames, ignore_index=True)
        df_mag['Bt'] = np.sqrt(df_mag['Bx']**2 + df_mag['By']**2 + df_mag['Bz']**2)

        bx = df_mag['Bx'].values
        by = df_mag['By'].values
        bz = df_mag['Bz'].values
        bt = df_mag['Bt'].values
        B_Mean = np.mean(bt)

        X = fft(bx)
        N = len(X)
        n = np.arange(N)
        sr = 8
        T = N / sr
        freq = n / T
        n_oneside = N / 2
        f_oneside = freq[:int(n_oneside)]
        psdx = ((np.abs(X[:int(n_oneside)])) ** 2) * 2 / N

        Y = fft(by)
        N = len(Y)
        n = np.arange(N)
        sr = 8
        T = N / sr
        freq = n / T
        n_oneside = N / 2
        f_oneside = freq[:int(n_oneside)]
        psdy = ((np.abs(Y[:int(n_oneside)])) ** 2) * 2 / N

        Z = fft(bz)
        N = len(Z)
        n = np.arange(N)
        sr = 8
        T = N / sr
        freq = n / T
        n_oneside = N / 2
        f_oneside = freq[:int(n_oneside)]
        psdz = ((np.abs(Z[:int(n_oneside)])) ** 2) * 2 / N

        bt = psdx + psdy + psdz
        n_oneside = N / 2
        f_oneside = freq[:int(n_oneside)]

        b_mean = float(B_Mean) * 10 ** -9
        q = 1.602176634 * 10 ** -19  # C
        mp = 1.6726219 * 10 ** -27  # kilograms
        Omega_cp = (q * b_mean) / mp  # rad/s
        f_cp = Omega_cp / (2 * 3.14)  # Hz
        va = b_mean / (mu_0 * n_p * mp) ** 0.5  # m/s

        f_pc = f_cp * (1 + v / va)  # Hz; proton cyclotron freq

        yz = (Y[:int(n_oneside)].conjugate()) * Z[:int(n_oneside)]
        YZ = 2 * (yz.imag)
        XYZ = abs(X[:int(n_oneside)]) ** 2 + abs(Y[:int(n_oneside)]) ** 2 + abs(Z[:int(n_oneside)]) ** 2
        sigma_m = YZ / XYZ

        fig = plt.figure(figsize=(10, 10))
        gs = GridSpec(nrows=2, ncols=1, height_ratios=[2,1])
        fig.tight_layout()
        a3 = fig.add_subplot(gs[1, 0])
        a4 = fig.add_subplot(gs[0,0])

        y_av = movingaverage(sigma_m, 960)
        y_av2 = movingaverage(sigma_m, 14400)

        zeroy = np.zeros((len(f_oneside), 1))

        a3.semilogx(f_oneside, y_av, label='Normailsed magnetic-helicity: 2min')
        a3.semilogx(f_oneside, y_av2, label='Normailsed magnetic-helicity: 30min')
        a3.legend(frameon=False, fontsize=10)
        a3.set_ylim([-0.4, 0.4])
        a3.plot(f_oneside, zeroy, color='black', ls='--')
        a3.set_xlabel(r'Frequency (Hz)', fontsize=14, fontweight='bold')
        a3.set_ylabel(r'$\sigma_m$', fontsize=14, fontweight='bold')
        a3.axvline(x=f_cp, color='black', ls='dotted')
        a3.axvline(x=f_pc, color='blue', ls='--')
        a3.axhline(y=0.5, color='b', ls='dotted')
        a3.axhline(y=-0.5, color='b', ls='dotted')

        y_av = movingaverage(bt, 10)
        a4.loglog(f_oneside, y_av, label='PSD of Bt')
        a4.legend(frameon=False, fontsize=10)
        a4.axvline(x=f_cp, color='black', ls='dotted')
        a4.axvline(x=f_pc, color='blue', ls='--')
        a4.set_ylabel(r'PSD ($nT^2/Hz$)', fontsize=14, fontweight='bold')
        a4.set_xlabel(r'Frequency (Hz)', fontsize=14, fontweight='bold')

        fontsize = 12
        for ax in [a3, a4]:
            for tick in ax.xaxis.get_major_ticks():
                tick.label1.set_fontsize(fontsize)
                tick.label1.set_fontweight('bold')
            for tick in ax.yaxis.get_major_ticks():
                tick.label1.set_fontsize(fontsize)
                tick.label1.set_fontweight('bold')

        start_date_timestamp = pd.to_datetime(start_date)
        plot_filename = f"{start_date_timestamp.strftime('%Y-%m-%d')}_plot.png"  
        plt.savefig(os.path.join(save_path, plot_filename), dpi=300)
```
